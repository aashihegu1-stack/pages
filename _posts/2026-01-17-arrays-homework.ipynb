```python
// Exercise 1: Array Basics
let favorites = ["Inception", "Minecraft", "The Hobbit", "Valorant", "Interstellar"];

console.log(favorites);
console.log(favorites[0]);
console.log(favorites[favorites.length - 1]);
console.log(favorites.length);

// Exercise 2: Modify Arrays
let shoppingList = ["milk", "eggs", "bread", "cheese"];

console.log(shoppingList);
shoppingList[1] = "butter";
shoppingList.push("yogurt");
shoppingList.splice(shoppingList.indexOf("bread"), 1);
console.log(shoppingList);

// Exercise 3: Loop Through an Array
let numbers = [10, 25, 30, 15, 20];
let sum = 0;

for (let i = 0; i < numbers.length; i++) {
    console.log("Number:", numbers[i]);
    console.log("Times 2:", numbers[i] * 2);
    sum += numbers[i];
}

console.log("Sum:", sum);
```
