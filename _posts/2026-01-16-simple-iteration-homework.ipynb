```python
// Simple array of numbers
let numbers = [2, 4, 6, 8, 10];

// Loop 1: for loop to print each number
for (let i = 0; i < numbers.length; i++) {
    console.log(numbers[i]);
}

// Loop 2: while loop to calculate the sum
let sum = 0;
let index = 0;

while (index < numbers.length) {
    sum += numbers[index];
    index++;
}

console.log("Total sum:", sum);

// Thank you Rishab for allowing late submission
```
