```python
%%js

// Problem #1: Check if a number is divisible by 5
let y = 25;

if (y % 5 === 0) {
    console.log(y + " is divisible by 5");
} else {
    console.log(y + " is not divisible by 5");
}

// Problem #2: Add three variables together
let a = 10;
let b = 15;
let sum = a + b;

console.log("a = " + a);
console.log("b = " + b);
console.log("sum = " + sum);
```
