**Task: Make a nested conditional that does the following:**
- contains more than 1 nested conditional
- use the numbers 1-50
- check if each number is divisible by each factor of 50

```python
// Factors of 50: 2, 5, 10, 25, 50
// Loop through numbers 1 to 50
for (let i = 1; i <= 50; i++) {
    if (i % 50 === 0) {
        console.log(i + " is divisible by 50 (divisible by 2, 5, 10, 25, and 50)");
    } else if (i % 25 === 0) {
        if (i % 10 === 0) {
            console.log(i + " is divisible by 25 and 10 (also by 2 and 5)");
        } else {
            console.log(i + " is divisible by 25 (also by 5)");
        }
    } else if (i % 10 === 0) {
        console.log(i + " is divisible by 10 (also by 2 and 5)");
    } else if (i % 5 === 0) {
        if (i % 2 === 0) {
            console.log(i + " is divisible by 5 and 2");
        } else {
            console.log(i + " is divisible by 5 only");
        }
    } else if (i % 2 === 0) {
        console.log(i + " is divisible by 2 only");
    }
}
```
